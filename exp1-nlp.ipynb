```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK data
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')

# Paragraph of about 45 words
text = """
Machine learning is helping students learn new concepts quickly.
Students are studying different algorithms and practicing them regularly.
They are learning from examples, solving problems, and improving their skills.
Teachers are also using technology to provide better learning experiences.
"""

#  TOKENIZATION

# Word Tokenization
words = word_tokenize(text)

print("WORD TOKENIZATION:")
print(words)

#  STEMMING

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in words]

print("\nSTEMMING:")
print(stemmed_words)


# LEMMATIZATION

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word.lower())
    for word in words
    if word.isalpha()
]

print("\nLEMMATIZATION:")
print(lemmatized_words)
```
